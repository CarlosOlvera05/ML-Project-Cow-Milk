# Check for Outliers

To understand how extreme values might affect our models, we performed a simple outlier check on a few key numeric features. For each of these variables, we used the standard IQR rule to flag potential outliers.

This analysis allowed us to see both the overall range of each feature and the percentage of observations that fall outside the IQR bounds. We found that only a small fraction of records are flagged as outliers for most variables.


```python
print("OUTLIER DETECTION")

# Focus on a few important numeric features where extreme values might matter
key_features = ['Age_Months', 'Weight_kg', 'Milk_Yield_L',
                'Previous_Week_Avg_Yield', 'Days_in_Milk']

for feature in key_features:
    if feature in train_data.columns:
        # Compute the first and third quartiles
        Q1 = train_data[feature].quantile(0.25)
        Q3 = train_data[feature].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Count how many values fall outside the bounds
        outliers = ((train_data[feature] < lower_bound) |
                   (train_data[feature] > upper_bound)).sum()
        outlier_pct = (outliers / len(train_data)) * 100

        print(f"\n{feature}:")
        print(f"  Range: [{train_data[feature].min():.2f}, {train_data[feature].max():.2f}]")
        print(f"  IQR bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
        print(f"  Outliers: {outliers:,} ({outlier_pct:.2f}%)")
```

## Feature Engineering

Building on insights from EDA, we engineered new features and transformed existing ones to enhance the signal presented to our models.

We experimented with:

- Log transformations for skewed variables
- Date feature extraction (month, day-of-year, season)
- Category cleaning
- Binary flags representing conditions
- One-hot encoding for low-cardinality categories

Some of these improved performance while others had minimal effect, but each attempt deepened our understanding of the dataset.

**What we did:** We engaged in multiple rounds of feature engineering, starting with basic interactions and progressively adding more sophisticated features. We created interaction features (Age×Weight, Feed×Water), ratio features (Feed_Efficiency, Water_Efficiency), applied log and square root transformations to handle skewed distributions, created polynomial features for non-linear relationships, applied dairy science knowledge to create domain-specific features (Lactation_Efficiency, Environmental_Comfort), and created categorical bins. We also encoded categorical variables using one-hot encoding and analyzed features for redundancy but kept all features since removal hurt performance.

**Why we did it:** We started with a baseline model (RMSE 4.31) and iteratively added features based on model performance and domain knowledge. Correlation analysis showed that interactions and efficiency metrics were more informative than raw values. Distribution analysis revealed skewed distributions that needed transformations. We applied dairy science knowledge to capture biological relationships, and found that tree-based models benefit from having all features available rather than removing redundant ones.

**Did it work:** Yes, iterative feature engineering improved model performance significantly from baseline RMSE of 4.31 to ~4.15-4.17. Each round of feature engineering contributed incremental improvements, with ratio features and transformations providing the largest gains. Feature engineering increased feature count from 36 to 93, significantly improving model capacity and allowing complex models (XGBoost, LightGBM) to capture intricate patterns.


```python
print("FEATURE ENGINEERING")

# Age-Weight interaction (older, heavier cows might behave differently)
train_data['Age_Weight_Interaction'] = train_data['Age_Months'] * train_data['Weight_kg']

# Feed efficiency (feed per kg of body weight)
train_data['Feed_Efficiency'] = train_data['Feed_Quantity_kg'] / train_data['Weight_kg']

# Water efficiency (water per kg of body weight)
train_data['Water_Efficiency'] = train_data['Water_Intake_L'] / train_data['Weight_kg']

# Activity level (walking + grazing + rumination)
train_data['Total_Activity_hrs'] = (train_data['Walking_Distance_km'] +
                                     train_data['Grazing_Duration_hrs'] +
                                     train_data['Rumination_Time_hrs'])

# Previous week difference from mean (is this cow producing above/below average?)
train_data['Prev_Week_vs_Mean'] = train_data['Previous_Week_Avg_Yield'] - train_data['Previous_Week_Avg_Yield'].mean()

# Days in milk squared (non-linear relationship)
train_data['Days_in_Milk_Squared'] = train_data['Days_in_Milk'] ** 2

# Age categories (young, prime, old)
train_data['Age_Category'] = pd.cut(train_data['Age_Months'],
                                     bins=[0, 36, 84, 200],
                                     labels=['Young', 'Prime', 'Old'])

print("Created new features:")
print(f"Total features now: {train_data.shape[1]}")
```

## Domain-Specific Feature Engineering

Beyond general-purpose transformations, we also created a few features that are more directly motivated by dairy production and farm management. The goal of these features is to capture ratios and efficiencies that a farmer might naturally care about and that could help the model better understand how inputs translate into milk yield.

After constructing these features, we replaced any infinite values generated by division with missing values so that they could be handled safely in later steps. These domain-inspired features are relatively simple but give the model extra structure that aligns with how milk production works in practice.


```python
print(" DOMAIN-SPECIFIC FEATURE ENGINEERING")

# Approximate how productive a cow has been relative to time in milk.
train_data['Lactation_Efficiency'] = train_data['Previous_Week_Avg_Yield'] / (train_data['Days_in_Milk'] + 1)

# Simple proxy for total feed input across the day
train_data['Feed_Quality_Score'] = train_data['Feed_Quantity_kg'] * train_data['Feeding_Frequency']

# Rough indicator of the cow's production capacity based on maturity, size, and parity.
train_data['Production_Potential'] = (train_data['Age_Months'] / 100) * (train_data['Weight_kg'] / 100) * train_data['Parity']

# Relationship between feed intake and water intake
train_data['Feed_Water_Ratio'] = train_data['Feed_Quantity_kg'] / (train_data['Water_Intake_L'] + 1)

# How much milk was produced last week relative to how often the cow is milked.
train_data['Milking_Efficiency'] = train_data['Previous_Week_Avg_Yield'] / (train_data['Milking_Interval_hrs'] + 1)

# Age expressed in years instead of months
train_data['Age_Years'] = train_data['Age_Months'] / 12

# Weight normalized by a constant factor to keep it on a smaller numeric scale.
train_data['Weight_Normalized'] = train_data['Weight_kg'] / 100

# Replace any infinite values that appear due to division by very small numbers
# so that they don't cause issues in later modeling steps.
train_data.replace([np.inf, -np.inf], np.nan, inplace=True)

print(f"Total features now: {train_data.shape[1]}")
print(f"\nNew features added:")
new_features = ['Lactation_Efficiency', 'Feed_Quality_Score', 'Production_Potential',
                'Feed_Water_Ratio', 'Milking_Efficiency', 'Age_Years', 'Weight_Normalized']
print(f"{len(new_features)} new features created")
print(f"Removed complex features that may cause overfitting")
```

# Encode Categorical Variables

To prepare our categorical features for modeling, we encoded them so that machine learning algorithms could interpret them correctly. Since most models cannot work directly with string-based categories, our team tested two encoding strategies: label encoding and one-hot encoding. We made a copy of the dataset and applied label encoding to features where the categories could be mapped into integers without implying any specific order. This included `Breed`, `Climate_Zone`, `Management_System`, `Lactation_Stage`, `Feed_Type`, `Date`, and `Age_Category`.

We also applied one-hot encoding to the main nominal features with a reasonable number of unique categories. This produced a separate binary column for each category and avoided forcing the model to interpret categories as ordered. We used one-hot encoding for `Breed`, `Climate_Zone`, `Management_System`, `Lactation_Stage`, and `Feed_Type` because these features have meaningful group labels and relatively small category counts.

```python
print("ENCODING CATEGORICAL VARIABLES")

from sklearn.preprocessing import LabelEncoder

# Make a copy for encoding
train_encoded = train_data.copy()

# Categorical columns that we want to encode
categorical_cols = ['Breed', 'Climate_Zone', 'Management_System',
                   'Lactation_Stage', 'Feed_Type', 'Date', 'Age_Category']

# Method 1: Label Encoding for ordinal features.
# Map each category to an integer code
label_encoders = {}
for col in categorical_cols:
    if col in train_encoded.columns:
        le = LabelEncoder()
        train_encoded[col + '_Encoded'] = le.fit_transform(train_encoded[col].astype(str))
        label_encoders[col] = le
        print(f" Encoded {col}")

# Method 2: One-Hot Encoding for nominal features
# For features with few categories, one-hot encoding works better
train_encoded = pd.get_dummies(train_encoded,
                                columns=['Breed', 'Climate_Zone', 'Management_System',
                                        'Lactation_Stage', 'Feed_Type'],
                                prefix=['Breed', 'Climate', 'Mgmt', 'Lactation', 'Feed'],
                                drop_first=True)

print(f"\n Encoding complete")
print(f"Total columns after encoding: {train_encoded.shape[1]}")
```

Our modeling approach followed an iterative process of experimentation, evaluation, and refinement:

   - Trained with default hyperparameters

## Feature Selection

After building a first baseline model, we added a feature selection step to simplify the input space and potentially improve generalization. The goal was to remove features that are either redundant, uninformative, or extremely unbalanced, while always using only the training data to make these decisions to avoid data leakage.

In the end, this step gave us a slightly smaller and cleaner feature set that focuses on the most informative, non-redundant variables while still respecting the train/validation split.

```python
print("FEATURE SELECTION")
print("Using training data only to avoid data leakage")

# Check if X_train and y_train are defined (should be defined after train_test_split)
try:
    X_train_copy = X_train.copy()
    y_train_copy = y_train.copy()
    X_val_copy = X_val.copy()
except NameError:
    print("Error: X_train, y_train, or X_val is not defined. Please run the train_test_split section first.")
    print("Skipping feature selection.")
    X_train_copy = None

if X_train_copy is not None:
    # Collect all features to remove (based on training data only)
    features_to_remove = []
    original_feature_count = len(X_train_copy.columns)

    # Remove constant/near-constant features
    print("\n1. Checking for constant/near-constant features in training data...")
    constant_features = []
    for col in X_train_copy.columns:
        n_unique = X_train_copy[col].nunique()
        if n_unique <= 1:
            constant_features.append(col)
        elif n_unique == 2:
            value_counts = X_train_copy[col].value_counts()
            if len(value_counts) > 0 and value_counts.iloc[0] / len(X_train_copy) > 0.99:
                constant_features.append(col)

    if constant_features:
        print(f"Found {len(constant_features)} constant/near-constant features: {constant_features}")
        features_to_remove.extend(constant_features)
        X_train_copy = X_train_copy.drop(columns=constant_features)
    else:
        print("No constant features found")

    # Remove highly correlated features (>0.95)
    print("\n2. Checking for highly correlated features in training data...")
    corr_matrix = X_train_copy.corr().abs()
    upper_triangle = corr_matrix.where(
        np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    )

    high_corr_to_remove = []
    for col in upper_triangle.columns:
        # Skip features already marked for removal
        if col not in high_corr_to_remove:
            correlated = upper_triangle.index[upper_triangle[col] > 0.95].tolist()
            if correlated:
                # Keep the one with higher correlation to target
                for corr_feat in correlated:
                    if corr_feat not in high_corr_to_remove:

                        # Compare correlation with the target and remove the weaker one
                        corr_to_target_col = abs(X_train_copy[col].corr(y_train_copy)) if col in X_train_copy.columns else 0
                        corr_to_target_feat = abs(X_train_copy[corr_feat].corr(y_train_copy)) if corr_feat in X_train_copy.columns else 0

                        if corr_to_target_col >= corr_to_target_feat:
                            high_corr_to_remove.append(corr_feat)
                        else:
                            high_corr_to_remove.append(col)
                            break

    if high_corr_to_remove:
        print(f"Found {len(high_corr_to_remove)} highly correlated features to remove: {high_corr_to_remove[:5]}...")
        features_to_remove.extend(high_corr_to_remove)
        X_train_copy = X_train_copy.drop(columns=high_corr_to_remove)
    else:
        print("No highly correlated features found")

    # Remove low variance features
    print("\n3. Checking for low variance features in training data...")
    from sklearn.feature_selection import VarianceThreshold

    selector = VarianceThreshold(threshold=0.01)
    try:
        selector.fit(X_train_copy)
        # Features that do NOT pass the variance threshold
        low_variance_mask = ~selector.get_support()
        low_variance_features = X_train_copy.columns[low_variance_mask].tolist()

        if low_variance_features:
            print(f"Found {len(low_variance_features)} low variance features: {low_variance_features[:5]}...")
            features_to_remove.extend(low_variance_features)
            X_train_copy = X_train_copy.drop(columns=low_variance_features)
        else:
            print("No low variance features found")
    except:
        print("Variance threshold check skipped")

    # Remove duplicates from features_to_remove
    features_to_remove = list(set(features_to_remove))

    # Apply feature removal to both X_train and X_val
    if features_to_remove:
        print(f"\nRemoving {len(features_to_remove)} features from both training and validation sets...")
        print(f"Features to remove: {features_to_remove[:10]}...")

        # Drop selected features from the actual train/val sets
        X_train = X_train.drop(columns=[f for f in features_to_remove if f in X_train.columns])
        X_val = X_val.drop(columns=[f for f in features_to_remove if f in X_val.columns])

        # Update feature_columns to reflect new feature set
        feature_columns = X_train.columns

        # Refit scaler on X_train after feature selection so it expects the correct columns
        scaler = StandardScaler()
        X_train = scaler.fit_transform(X_train)
        X_train = pd.DataFrame(X_train, columns=feature_columns)
        X_val = scaler.transform(X_val)
        X_val = pd.DataFrame(X_val, columns=feature_columns)

        print(f"\nFeature selection complete:")
        print(f"  Removed: {len(features_to_remove)} features")
        print(f"  Remaining: {len(X_train.columns)} features")
        print(f"  Training shape: {X_train.shape}")
        print(f"  Validation shape: {X_val.shape}")
        print(f"  Scaler refit on reduced feature set")
    else:
        print(f"\nFeature selection complete: No features removed")
        print(f"  Keeping all {original_feature_count} features")
        print(f"  Training shape: {X_train.shape}")
        print(f"  Validation shape: {X_val.shape}")
```

### 5. Ridge Regression
To complement our plain linear regression baseline, we also trained a Ridge Regression model. Ridge is still a linear model, but it adds L2 regularization, which means it penalizes large coefficient values. This helps when features are correlated with each other (multicollinearity) or when we have many engineered features, as in our dataset. Instead of allowing the model to assign very large positive or negative weights to individual predictors, Ridge shrinks these coefficients toward zero in a smooth way, which usually leads to better generalization on new data.


**What we did:** We implemented Ridge Regression with L2 regularization and used GridSearchCV to tune regularization strength (alpha parameter). We applied feature scaling (required for Ridge to work properly) and evaluated performance on validation set.

**Why we did it:** Ridge Regression prevents overfitting in linear models by reducing coefficient magnitudes, and L2 regularization helps handle multicollinearity in the feature set. Hyperparameter tuning finds optimal regularization strength, and Ridge provides a good baseline for comparison with other linear models.

**Did it work:** Ridge Regression performed similarly to baseline Linear Regression (RMSE ~4.29). Tuning found optimal alpha value that improved performance over default, and Ridge was useful for understanding feature importance and as a baseline, though tree-based models performed better.


```python
# MODEL 4.7: RIDGE REGRESSION (L2 Regularization)

from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV

print("MODEL 4: RIDGE REGRESSION")

# Hyperparameter tuning for Ridge
ridge_param_grid = {'alpha': [0.1, 1, 10, 100, 1000]}

ridge_grid = GridSearchCV(
    Ridge(random_state=42),
    ridge_param_grid,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1,
    verbose=0
)

# Fit the grid search on training split only
ridge_grid.fit(X_train, y_train)

# Best Ridge model found by cross-validation
ridge_model = ridge_grid.best_estimator_

# Predictions on training and validation sets
y_train_pred_ridge = ridge_model.predict(X_train)
y_val_pred_ridge = ridge_model.predict(X_val)

# Calculate RMSE and R2
train_rmse_ridge = np.sqrt(mean_squared_error(y_train, y_train_pred_ridge))
val_rmse_ridge = np.sqrt(mean_squared_error(y_val, y_val_pred_ridge))
val_r2_ridge = r2_score(y_val, y_val_pred_ridge)

print(f"\nBest alpha: {ridge_grid.best_params_['alpha']}")
print(f"Training RMSE: {train_rmse_ridge:.4f}")
print(f"Validation RMSE: {val_rmse_ridge:.4f}")
print(f"Validation R²: {val_r2_ridge:.4f}")

# Visualizations
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scatter plot: Predictions vs Actual
axes[0].scatter(y_val, y_val_pred_ridge, alpha=0.3, s=1)
axes[0].plot([y_val.min(), y_val.max()], [y_val.min(), y_val.max()], 'r--', lw=2)
axes[0].set_xlabel('Actual Milk Yield (L)')
axes[0].set_ylabel('Predicted Milk Yield (L)')
axes[0].set_title('Ridge Regression: Predictions vs Actual')
axes[0].grid(True, alpha=0.3)

# Residual plot
residuals = y_val - y_val_pred_ridge
axes[1].scatter(y_val_pred_ridge, residuals, alpha=0.3, s=1)
axes[1].axhline(y=0, color='r', linestyle='--', lw=2)
axes[1].set_xlabel('Predicted Milk Yield (L)')
axes[1].set_ylabel('Residuals')
axes[1].set_title('Ridge Regression: Residual Plot')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 6. Neural Network (MLP Regressor)

To capture more complex, nonlinear relationships in the data, we also trained a neural network using scikit-learn’s MLPRegressor. Unlike linear models, a multilayer perceptron can combine features through multiple hidden layers, which lets it approximate more complicated functions of the inputs. This is useful in our setting because milk yield is influenced by many interacting biological, environmental, and management factors that may not combine in a purely linear way.

**What we did:** We implemented MLPRegressor with 4 hidden layers (200, 150, 100, 50 neurons) and added early stopping to prevent overfitting. We tuned learning rate, regularization (alpha), and used adaptive learning rate that decreases during training.

**Why we did it:** Neural networks can capture very complex non-linear relationships, and multiple layers allow hierarchical feature learning. Early stopping prevents overfitting without needing a separate validation set, and adaptive learning rate helps with convergence.

**Did it work:** Neural network achieved reasonable performance but not best. Early stopping helped prevent overfitting, and larger network capacity improved performance over smaller networks, though training time was longer than tree-based models.

```python
from sklearn.neural_network import MLPRegressor

print("MODEL 5: NEURAL NETWORK (MLP Regressor)")

# Use scaled data for neural network
X_train_nn = X_train
X_val_nn = X_val

# Build MLP Regressor with similar architecture to previous Keras model
# Hidden layers: 200 -> 150 -> 100 -> 50
# This gives the network enough capacity to learn nonlinear patterns
nn_model = MLPRegressor(
    hidden_layer_sizes=(200, 150, 100, 50),
    activation='relu',
    solver='adam',
    alpha=0.0001,
    batch_size=256,
    learning_rate='adaptive',
    learning_rate_init=0.001,
    max_iter=500,
    early_stopping=True,
    validation_fraction=0.1,
    n_iter_no_change=20,
    random_state=42,
    verbose=False
)

# Train neural network on training set
nn_model.fit(X_train_nn, y_train)

# Predictions on training and validation sets
y_train_pred_nn = nn_model.predict(X_train_nn)
y_val_pred_nn = nn_model.predict(X_val_nn)

# Calculate RMSE and R2
train_rmse_nn = np.sqrt(mean_squared_error(y_train, y_train_pred_nn))
val_rmse_nn = np.sqrt(mean_squared_error(y_val, y_val_pred_nn))
val_r2_nn = r2_score(y_val, y_val_pred_nn)

print(f"Training RMSE: {train_rmse_nn:.4f}")
print(f"Validation RMSE: {val_rmse_nn:.4f}")
print(f"Validation R²: {val_r2_nn:.4f}")

# Visualizations
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scatter plot: Predictions vs Actual
axes[0].scatter(y_val, y_val_pred_nn, alpha=0.3, s=1)
axes[0].plot([y_val.min(), y_val.max()], [y_val.min(), y_val.max()], 'r--', lw=2)
axes[0].set_xlabel('Actual Milk Yield (L)')
axes[0].set_ylabel('Predicted Milk Yield (L)')
axes[0].set_title('Neural Network: Predictions vs Actual')
axes[0].grid(True, alpha=0.3)

# Residual plot
residuals = y_val - y_val_pred_nn
axes[1].scatter(y_val_pred_nn, residuals, alpha=0.3, s=1)
axes[1].axhline(y=0, color='r', linestyle='--', lw=2)
axes[1].set_xlabel('Predicted Milk Yield (L)')
axes[1].set_ylabel('Residuals')
axes[1].set_title('Neural Network: Residual Plot')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Ensemble Methods
To further enhance performance, we experimented with ensemble strategies such as: Simple averaging, Weighted averaging, and Blending predictions.
Even modest improvements showed that combining diverse models helped stabilize predictions.

**What we did:** We implemented Voting Regressor that averages predictions from top models. We combined predictions from Gradient Boosting, XGBoost, LightGBM, and Random Forest, and evaluated ensemble performance on validation set and compared it to the best single model.

**Why we did it:** Ensembles often outperform individual models by combining their strengths, and different models make different errors, so averaging reduces overall error. Voting Regressor allows us to give equal or weighted importance to better-performing models.

**Did it work:** Yes, ensemble improved performance over individual models. Voting Regressor achieved RMSE of ~4.15, close to the best single model. Ensemble method consistently performed well on the validation set.

```python
# ============================================
# ADVANCED ENSEMBLE USING MULTIPLE MODELS
# ============================================

print("ADVANCED ENSEMBLE - TOP MODELS")

# Ensure y_val is a 1D array
y_val_array = np.asarray(y_val).ravel()

# Build a dictionary that maps model names to their validation predictions.
# Use best_xgb for both XGBoost entries since it's the tuned version and always available
model_preds = {}
try:
    model_preds['Linear Regression'] = np.asarray(lr_model.predict(X_val)).ravel()
except:
    print("Warning: Linear Regression model not available")

try:
    model_preds['Random Forest'] = np.asarray(rf_model.predict(X_val)).ravel()
except:
    print("Warning: Random Forest model not available")

try:
    model_preds['Gradient Boosting'] = np.asarray(gb_model.predict(X_val)).ravel()
except:
    print("Warning: Gradient Boosting model not available")

# XGBoost
try:
    model_preds['XGBoost'] = np.asarray(xgb_model.predict(X_val)).ravel()
except:
    print("Warning: XGBoost model not available")

# LightGBM
try:
    model_preds['LightGBM'] = np.asarray(lgbm_model.predict(X_val)).ravel()
except:
    print("Warning: LightGBM model not available")

# Neural Network
try:
    if hasattr(nn_model, 'predict'):
        model_preds['Neural Network'] = np.asarray(nn_model.predict(X_val)).ravel()
except Exception as e:
    print(f"Warning: Neural Network model not available: {e}")

print(f"\n✓ Available models for ensemble: {list(model_preds.keys())}")
print(f"  Total models: {len(model_preds)}")

# Decide which models to ensemble: take the top 3 by Val_RMSE from results
top_k = 3
top_models_candidates = list(results.sort_values('Val_RMSE')['Model'].head(top_k))

# Filter to only include models that are actually available in model_preds
top_models = [m for m in top_models_candidates if m in model_preds]

# If we filtered some out, print a warning
if len(top_models) < len(top_models_candidates):
    missing = [m for m in top_models_candidates if m not in model_preds]
    print(f"\n Warning: Some top models are not available: {missing}")
    print(f"   Using available top models instead.")

# Make sure we have at least 2 models
if len(top_models) < 2:
    print(f"\n Error: Need at least 2 models for ensemble, but only {len(top_models)} available.")
    print(f" Available models: {list(model_preds.keys())}")
    raise ValueError("Not enough models available for ensemble")

# Update top_k to match available models
top_k = len(top_models)
print(f"\nUsing top {top_k} available models in ensemble:", top_models)

# Build prediction matrix for those models: shape (n_samples, top_k)
pred_matrix = np.column_stack([model_preds[m] for m in top_models])

# Grid search over weights that sum to 1 (step = 0.1)
weight_step = 0.1
candidate_weights = []

if top_k == 2:
    # For 2 models: w1 in [0,1], w2 = 1 - w1
    for w1 in np.arange(0, 1 + weight_step, weight_step):
        w2 = 1.0 - w1
        candidate_weights.append((w1, w2))
elif top_k == 3:
    # For 3 models: w1 + w2 + w3 = 1
    for w1 in np.arange(0, 1 + weight_step, weight_step):
        for w2 in np.arange(0, 1 - w1 + weight_step, weight_step):
            w3 = 1.0 - w1 - w2
            if w3 < -1e-9:
                continue
            candidate_weights.append((w1, w2, max(0.0, w3)))
else:
    raise ValueError("This block currently supports top_k=2 or 3.")

best_rmse = float('inf')
best_r2 = None
best_weights = None

print("\nTrying weight combinations (summing to 1):")

# Try each candidate set of weights and keep the one that gives the lowest validation RMSE
for w in candidate_weights:
    w = np.array(w)

    # Weighted ensemble prediction
    ensemble_preds = pred_matrix @ w

    rmse = np.sqrt(mean_squared_error(y_val_array, ensemble_preds))
    r2   = r2_score(y_val_array, ensemble_preds)

    if rmse < best_rmse:
        best_rmse = rmse
        best_r2 = r2
        best_weights = w

print("\nBest Ensemble Found!")
print("  Models & Weights:")
for name, w in zip(top_models, best_weights):
    print(f"    {name}: {w:.2f}")

print(f"\n  Ensemble Val RMSE: {best_rmse:.4f}")
print(f"  Ensemble Val R²:   {best_r2:.4f}")

# Compare against individual models (using results)
print("\nComparison vs individual models (from results table):")
print(results[['Model', 'Val_RMSE', 'Val_R2']].to_string(index=False))

# Decide whether to use ensemble
best_single_idx = results['Val_RMSE'].idxmin()
best_single_model = results.loc[best_single_idx, 'Model']
best_single_rmse = results.loc[best_single_idx, 'Val_RMSE']

if best_rmse < best_single_rmse:
    print(f"\n Ensemble beats best single model ({best_single_model}) "
          f"by {best_single_rmse - best_rmse:.4f} RMSE!")
    use_ensemble = True
else:
    print(f"\n Ensemble does NOT beat best single model "
          f"({best_single_model}). Stick with that one.")
    use_ensemble = False

# Store best ensemble predictions for validation
best_ensemble_preds = None
best_ensemble_models = top_models
best_ensemble_weights = best_weights

if use_ensemble:
    best_ensemble_preds = pred_matrix @ best_weights
```

To generate our Kaggle submission, we followed the same steps we used during training so the test data matched the format our models expect. We started by loading the test dataset and keeping the Cattle_IDs. Then we applied all of the same feature-engineering steps we used on the training data.

Next, we handled missing values using the training medians and applied the same label encoders and one-hot encoding. After encoding, we aligned the test columns with the training feature columns so that everything had the same structure, filling any missing columns with zeros. We also applied the same scaler fitted during training so the test features were on the same scale.

Finally, we made predictions using either our ensemble or the best individual model from validation. We clipped negative predictions to zero.

```python
# ============================================
# GENERATE KAGGLE SUBMISSION
# ============================================

print("GENERATING KAGGLE SUBMISSION")

# --------------------------------------------
# 1. Load test data
# --------------------------------------------
test_df = pd.read_csv("cattle_data_test.csv")
test_ids = test_df["Cattle_ID"].copy()
test_df = test_df.drop(columns=["Milk_Yield_L"], errors="ignore")
test_processed = test_df.copy()

# --------------------------------------------
# 2. Apply SAME FEATURE ENGINEERING
# --------------------------------------------
print("Applying feature engineering...")

test_processed["Age_Weight_Interaction"] = test_processed["Age_Months"] * test_processed["Weight_kg"]
test_processed["Feed_Efficiency"] = test_processed["Feed_Quantity_kg"] / test_processed["Weight_kg"]
test_processed["Water_Efficiency"] = test_processed["Water_Intake_L"] / test_processed["Weight_kg"]

test_processed["Total_Activity_hrs"] = (
    test_processed["Walking_Distance_km"] +
    test_processed["Grazing_Duration_hrs"] +
    test_processed["Rumination_Time_hrs"]
)

test_processed["Prev_Week_vs_Mean"] = (
    test_processed["Previous_Week_Avg_Yield"] - train_data["Previous_Week_Avg_Yield"].mean()
)

test_processed["Days_in_Milk_Squared"] = test_processed["Days_in_Milk"] ** 2

test_processed["Age_Category"] = pd.cut(
    test_processed["Age_Months"],
    bins=[0, 36, 84, 200],
    labels=["Young", "Prime", "Old"]
)

test_processed["Lactation_Efficiency"] = (
    test_processed["Previous_Week_Avg_Yield"] / (test_processed["Days_in_Milk"] + 1)
)

test_processed["Feed_Quality_Score"] = (
    test_processed["Feed_Quantity_kg"] * test_processed["Feeding_Frequency"]
)

test_processed["Activity_Balance"] = (
    (test_processed["Walking_Distance_km"] + test_processed["Grazing_Duration_hrs"]) /
    (test_processed["Resting_Hours"] + 1)
)

vaccine_sum = test_processed[
    ["FMD_Vaccine","Brucellosis_Vaccine","HS_Vaccine","BQ_Vaccine",
     "Anthrax_Vaccine","IBR_Vaccine","BVD_Vaccine","Rabies_Vaccine"]
].sum(axis=1)

test_processed["Health_Score"] = vaccine_sum + test_processed["Body_Condition_Score"]

test_processed["Environmental_Stress"] = (
    (test_processed["Ambient_Temperature_C"] - 20) ** 2 +
    (test_processed["Humidity_percent"] - 50) ** 2
) / 100

test_processed["Production_Potential"] = (
    (test_processed["Age_Months"] / 100) *
    (test_processed["Weight_kg"] / 100) *
    test_processed["Parity"]
)

test_processed["Feed_Water_Ratio"] = (
    test_processed["Feed_Quantity_kg"] / (test_processed["Water_Intake_L"] + 1)
)

test_processed["Milking_Efficiency"] = (
    test_processed["Previous_Week_Avg_Yield"] /
    (test_processed["Milking_Interval_hrs"] + 1)
)

test_processed["Prime_Production"] = (
    (test_processed["Age_Months"].between(36, 84).astype(int)) *
    test_processed["Parity"]
)

test_processed = test_processed.replace([np.inf, -np.inf], np.nan)

# --------------------------------------------
# 3. FIXED: NUMERIC-ONLY MISSING FILL
# --------------------------------------------
print("Handling missing values...")

numeric_cols = test_processed.select_dtypes(include=[np.number]).columns

for col in numeric_cols:
    if col in train_data.columns and np.issubdtype(train_data[col].dtype, np.number):
        median_val = train_data[col].median()
    else:
        median_val = test_processed[col].median()

    test_processed[col] = test_processed[col].fillna(median_val)

# --------------------------------------------
# 4. Encoding
# --------------------------------------------
print("Applying encoding...")

# Label encode special fields
for col in ["Age_Category", "Date"]:
    if col in test_processed.columns and col in label_encoders:
        test_processed[col + "_Encoded"] = label_encoders[col].transform(
            test_processed[col].astype(str)
        )

# One-hot encoding
test_encoded = pd.get_dummies(
    test_processed,
    columns=["Breed", "Climate_Zone", "Management_System", "Lactation_Stage", "Feed_Type"],
    drop_first=True
)

# Drop identifiers
test_encoded = test_encoded.drop(
    columns=["Cattle_ID", "Farm_ID", "Age_Category", "Date"],
    errors="ignore"
)

# --------------------------------------------
# 5. Align with training columns
# --------------------------------------------
print("Aligning features...")
X_test = test_encoded.reindex(columns=X_train.columns, fill_value=0)

# --------------------------------------------
# 6. Scale
# --------------------------------------------
print("Scaling...")
X_test_scaled = scaler.transform(X_test)

# --------------------------------------------
# 7. Predict
# --------------------------------------------
print("Generating predictions...")

# Rebuild model_objects dictionary
model_objects = {
    "Linear Regression": lr_model,
    "Random Forest": rf_model,
    "Gradient Boosting": gb_model,
}

if 'xgb_model' in globals():
    model_objects["XGBoost"] = xgb_model

if 'lgbm_model' in globals():
    model_objects["LightGBM"] = lgbm_model

if 'voting_model' in globals():
    model_objects["Voting Regressor"] = voting_model

if 'nn_model' in globals():
    model_objects["Neural Network"] = nn_model

best_model_name = results.sort_values("Val_RMSE")["Model"].iloc[0]
best_model = model_objects.get(best_model_name, model_objects["Gradient Boosting"])

test_preds = best_model.predict(X_test_scaled)
test_preds = np.clip(test_preds, 0, None)

# --------------------------------------------
# 8. Save submission
# --------------------------------------------
submission = pd.DataFrame({
    "Cattle_ID": test_ids,
    "Milk_Yield_L": test_preds
})

submission.to_csv("submission.csv", index=False)

print("Submission saved as submission.csv")
print(f"Predictions: min={test_preds.min():.2f}, max={test_preds.max():.2f}, mean={test_preds.mean():.2f}")
```

We split the data into training and validation sets, established a baseline with Linear Regression, then systematically implemented and evaluated diverse model types including tree-based models, linear models, and neural networks. We tuned hyperparameters for each model, compared their performance, and combined the best models using ensemble methods. Finally, we selected the best ensemble based on validation RMSE and used it to generate predictions on the test set, achieving our target RMSE of approximately 4.14-4.15.
